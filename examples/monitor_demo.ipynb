# peetsfea monitor demo

This notebook launches the monitor UI with machine rows and process tables.

```python
import peetsfea
import peetsfea.config as peetsfea_config
from peetsfea.types import AedtMachine, PyaedtProcessInfo

this_pc: AedtMachine = AedtMachine(
    name="this_pc",
    is_current_pc=True,
    ip_address="localhost",
    aedt_version="v242",
)

processes = [
    PyaedtProcessInfo(
        status="running",
        gui_enabled=False,
        stage="solve",
        stage_elapsed_min=12.4,
        design_name="ipt_demo",
        toml_path="examples/ipt_demo.toml",
        aedt_version="v242",
        ip_address="localhost",
    ),
    PyaedtProcessInfo(
        status="queued",
        gui_enabled=False,
        stage="startup",
        stage_elapsed_min=2.0,
        design_name="ipt_sweep",
        toml_path="examples/ipt_sweep.toml",
        aedt_version="v242",
        ip_address="localhost",
    ),
]

peetsfea_config.machine_list.append(this_pc)
peetsfea_config.pyaedt_processes_by_machine[this_pc.name] = processes

peetsfea.start_monitor()
```
